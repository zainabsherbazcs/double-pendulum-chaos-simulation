```python
import numpy as np
import matplotlib.pyplot as plt

# Let's define the physics parameters of our double pendulum
L1, L2 = 1.0, 1.0  # Length of the arms (in meters)
m1, m2 = 1.0, 1.0  # Mass of the bobs (in kilograms)
g = 9.81           # Acceleration due to gravity (m/s^2)

print("Physics simulation libraries imported successfully! Ready for equations.")
```

```python
from scipy.integrate import solve_ivp

# Define the differential equations governing the double pendulum
def equations(t, y, L1, L2, m1, m2, g):
    theta1, z1, theta2, z2 = y

    c, s = np.cos(theta1 - theta2), np.sin(theta1 - theta2)

    # Mathematical denominators for the acceleration formulas
    den1 = L1 * (2 * m1 + m2 - m2 * np.cos(2 * theta1 - 2 * theta2))
    den2 = L2 * (2 * m1 + m2 - m2 * np.cos(2 * theta1 - 2 * theta2))

    # Angular acceleration equations derived using Lagrangian mechanics
    theta1_dot_dot = (-g * (2 * m1 + m2) * np.sin(theta1) - m2 * g * np.sin(theta1 - 2 * theta2) -
                      2 * s * m2 * (z2**2 * L2 + z1**2 * L1 * c)) / den1

    theta2_dot_dot = (2 * s * (z1**2 * L1 * (m1 + m2) + g * (m1 + m2) * np.cos(theta1) +
                      z2**2 * L2 * m2 * c)) / den2

    return [z1, theta1_dot_dot, z2, theta2_dot_dot]

print("Physics equations mapped successfully! Ready to simulate.")
```

```python
# Time span for the simulation: 0 to 15 seconds, tracking 1500 individual moments
t_span = (0, 15)
t_eval = np.linspace(0, 15, 1500)

# Pendulum 1: Starts exactly horizontal (90 degrees for both arms)
initial_state1 = [np.radians(90.0), 0.0, np.radians(90.0), 0.0]

# Pendulum 2: Starts at 90.001 degrees (A tiny change of 0.001 degrees!)
initial_state2 = [np.radians(90.001), 0.0, np.radians(90.0), 0.0]

# Run the computer algorithms to solve the math over time
solution1 = solve_ivp(equations, t_span, initial_state1, args=(L1, L2, m1, m2, g), t_eval=t_eval)
solution2 = solve_ivp(equations, t_span, initial_state2, args=(L1, L2, m1, m2, g), t_eval=t_eval)

# Convert the mathematical angles into (X, Y) coordinates so we can map them
x2_1 = L1 * np.sin(solution1.y[0]) + L2 * np.sin(solution1.y[2])
y2_1 = -L1 * np.cos(solution1.y[0]) - L2 * np.cos(solution1.y[2])

x2_2 = L1 * np.sin(solution2.y[0]) + L2 * np.sin(solution2.y[2])
y2_2 = -L1 * np.cos(solution2.y[0]) - L2 * np.cos(solution2.y[2])

print("Simulation finished processing! Ready to plot the chaos.")
```

```python
# Create the figure workspace
plt.figure(figsize=(12, 6))

# Plot the trajectory of Pendulum 1 (Base Case)
plt.plot(t_eval, np.degrees(solution1.y[2]), label='Pendulum 1 (Starts at 90.0°)', color='blue', alpha=0.8)

# Plot the trajectory of Pendulum 2 (Altered by 0.001°)
plt.plot(t_eval, np.degrees(solution2.y[2]), label='Pendulum 2 (Starts at 90.001°)', color='red', linestyle='--', alpha=0.8)

# Add academic titles and labels for your paper
plt.title('Sensitivity to Initial Conditions: The Visual Proof of Chaos Theory', fontsize=14, fontweight='bold')
plt.xlabel('Time (Seconds)', fontsize=12)
plt.ylabel('Angle of Lower Bob (Degrees)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)

# Display the final research plot
plt.show()
```

```python
# Calculate the absolute difference between the angles over time
angle_difference = np.abs(solution1.y[2] - solution2.y[2])

# Avoid log(0) errors by filtering out identical values at the start
valid_indices = angle_difference > 1e-12
t_valid = t_eval[valid_indices]
log_diff = np.log(angle_difference[valid_indices])

# Use linear regression on the phase before total saturation (from 0 to 9.6 seconds)
analysis_limit = t_valid < 9.6
slope, intercept = np.polyfit(t_valid[analysis_limit], log_diff[analysis_limit], 1)

print(f"Calculated Lyapunov Exponent (λ): {slope:.4f}")
if slope > 0:
    print("Mathematical Proof: The positive exponent firmly confirms deterministic chaos!")
else:
    print("The system is stable.")
```
